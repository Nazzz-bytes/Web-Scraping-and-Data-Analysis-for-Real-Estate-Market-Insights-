```python
# Imports
import requests
from bs4 import BeautifulSoup
import pandas as pd

# List of URLs to scrape
urls = [
    "https://www.redfin.com/WA/Seattle/9819-15th-Ave-NE-98115/home/194627117",
    "https://www.redfin.com/WA/Seattle/1400-N-80th-St-98103/home/21884185?utm_campaign=copy&utm_medium=share&utm_source=web_share&s_id=itaHmqfeTiyNSZMDPA8w-w",
    "https://redf.in/GJ1ZHY",
    "https://www.redfin.com/WA/Seattle/116-17th-Ave-98122/unit-D/home/52563?utm_campaign=copy&utm_medium=share&utm_source=web_share&s_id=itaHmqfeTiyNSZMDPA8w-w",
    "https://www.redfin.com/WA/Seattle/154-35th-Ave-E-98112/home/140639?utm_campaign=copy&utm_medium=share&utm_source=web_share&s_id=itaHmqfeTiyNSZMDPA8w-w",
    "https://redf.in/VLckDn",
    "https://www.redfin.com/WA/Burien/10915-Glen-Acres-Dr-S-98168/unit-21A/home/175114471?utm_campaign=copy&utm_medium=share&utm_source=web_share&s_id=itaHmqfeTiyNSZMDPA8w-w",
    "https://redf.in/Ja8KTa",
    "https://www.redfin.com/WA/Seattle/921-W-McGraw-St-98119/unit-34/home/169507123?utm_campaign=copy&utm_medium=share&utm_source=web_share&s_id=itaHmqfeTiyNSZMDPA8w-w",
    "https://www.redfin.com/WA/Seattle/9819-15th-Ave-NE-98115/home/194627117?utm_campaign=copy&utm_medium=share&utm_source=web_share&s_id=itaHmqfeTiyNSZMDPA8w-w",
    "https://www.redfin.com/WA/Seattle/3008-NW-85th-St-98117/home/164112?utm_campaign=copy&utm_medium=share&utm_source=web_share&s_id=itaHmqfeTiyNSZMDPA8w-w",
    "https://redf.in/Teqait",
]

# Define headers
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# List to store the data
data = []

# Loop through each URL with an index
for i, url in enumerate(urls, start=1):
    print(f"Processing {i} of {len(urls)}")

    # Send an HTTP request to fetch the page content
    response = requests.get(url, headers=headers)

    # Raise exception if request was unsuccessful
    if response.status_code != 200:
        raise Exception(f"Failed to retrieve data from {url}. \nStatus code: {response.status_code}, \nError: {response.text}")

    # Parse the HTML content of the page using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # Extract address
    address_tag = soup.find("div", class_="street-address", attrs={"data-rf-test-id": "abp-streetLine"})
    address = address_tag.get_text(strip=True) if address_tag else "N/A"

    # Extract price (first statsValue usually corresponds to price)
    price_tag = soup.find("div", class_="statsValue")
    price = price_tag.get_text(strip=True) if price_tag else "N/A"

    # Extract beds
    beds_section = soup.find("div", class_="stat-block beds-section", attrs={"data-rf-test-id": "abp-beds"})
    beds = beds_section.find("div", class_="statsValue").get_text(strip=True) if beds_section else "N/A"

    # Extract baths
    baths_section = soup.find("div", class_="stat-block baths-section", attrs={"data-rf-test-id": "abp-baths"})
    baths = baths_section.find("div", class_="statsValue").get_text(strip=True) if baths_section else "N/A"

    # Extract downpayment
    downpayment_tag = soup.find("span", class_="value-content")
    downpayment = downpayment_tag.get_text(strip=True) if downpayment_tag else "N/A"

    # Append the extracted data to the list
    data.append({
        "address": address,
        "beds": beds,
        "baths": baths,
        "price": price,
        
    })

# Create a DataFrame from the list of dictionaries
df = pd.DataFrame(data)

# Display the DataFrame
print(df)

# Save to CSV
df.to_csv('redfin_home_data.csv', index=False)
```
